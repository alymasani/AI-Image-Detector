```python
# Reads whatever runs scripts/train.py has written, so new models and seeds show up here.
import pandas as pd
from IPython.display import display

from detector.metrics import read_predictions, summarize
from detector.plots import (
    TEST_SETS,
    plot_generalization_gap,
    plot_roc_curves,
    plot_training_curves,
    save_figure,
)
from detector.training import RESULTS_DIR

metrics = pd.read_csv(RESULTS_DIR / "metrics.csv")
display(summarize(metrics))

save_figure(plot_generalization_gap(metrics), "generalization_gap")

# ROC curves from each model's lowest seed.
first_runs = metrics.sort_values("seed").groupby("model")["run"].first()
curves = {
    model: {t: read_predictions(RESULTS_DIR / "predictions" / f"{run}__{t}.csv") for t in TEST_SETS}
    for model, run in first_runs.items()
}
save_figure(plot_roc_curves(curves), "roc_curves")

histories = {
    run: pd.read_csv(RESULTS_DIR / "runs" / run / "history.csv") for run in metrics["run"].unique()
}
save_figure(plot_training_curves(histories), "training_curves")
```
